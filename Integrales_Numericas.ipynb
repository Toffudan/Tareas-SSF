Primero definimos las librerias y las funciones a evaluar

```python
import math
import sympy as sp
import numpy as np

Funciones = [
    "((math.e)**((-x**2)/2))/(math.sqrt(2*math.pi))",
    "(((math.e**x)*math.sin(x))/(1+(x**2)))","((math.e**x)/x)","((1-(math.e**x)/1))"]
```

Despues definimos completamente las formas de integracion numerica

```python
def Int_Trap(f,b,a,n):
    h = (b-a)/float(n)
    I = (f(a)+f(b))/2
    for k in range(1,n):
        x = a + k*h
        I += f(x)
    return I*h


def Int_Simpson(f,b,a,n):

  h=(b-a)/float(n)
  I= (f(a)+f(b))

  for k in range(1,n):
    x=a+ k*h
    if k % 2 == 0:
      I += 2 * f(x)
    else:
      I += 4 * f(x)
  return I*(h/3)


def gauss_legendre(f, a, b, n):
    x, w = np.polynomial.legendre.leggauss(n)
    xm = 0.5 * (b + a)
    xr = 0.5 * (b - a)
    I = 0
    for i in range(n):
        I += w[i] * f(xm + xr * x[i])
    return I * abs(xr) * (1 if b <= a else -1)
```

Definimos una funcion para no tener indeterminaciones de divicion entre 0

```python
def safe_eval(f, x):
    try:
        return f(x)
    except ZeroDivisionError:
        return 0
```

Usamos un ciclo en el que decimos cual elemento del arreglo de funciones se va a derivar y usamos las definiciones que tenemos de las integrales, a su vez hacemos que los elementos de este pasen tengan una forma simbolica con el sympy, para poder posteriormente realizar las derivadas para el calculo de error. Por ultimo imprimimos los resultados.

```python
import math
import sympy as sp
import numpy as np

Funciones = [
    "((math.e)**((-x**2)/2))/(math.sqrt(2*math.pi))",
    "(((math.e**x)*math.sin(x))/(1+(x**2)))","((math.e**x)/x)","((1-(math.e**x)/1))"]

def safe_eval(f, x):
    try:
        return f(x)
    except ZeroDivisionError:
        return 0

def Int_Trap(f,b,a,n):
    h = (b-a)/float(n)
    I = (f(a)+f(b))/2
    for k in range(1,n):
        x = a + k*h
        I += safe_eval(f, x)
    return I*h


def Int_Simpson(f,b,a,n):

  h=(b-a)/float(n)
  I= (f(a)+f(b))

  for k in range(1,n):
    x=a+ k*h
    if k % 2 == 0:
      I += 2 * safe_eval(f, x)
    else:
      I += 4 * safe_eval(f, x)
  return I*(h/3)


def gauss_legendre(f, a, b, n):
    x, w = np.polynomial.legendre.leggauss(n)
    xm = 0.5 * (b + a)
    xr = 0.5 * (b - a)
    I = 0
    for i in range(n):
        I += w[i] * f(xm + xr * x[i])
    return I * abs(xr) * (1 if b <= a else -1)


for z in range(len(Funciones)):
    expr_str = Funciones[z].replace("math.e", "E").replace("math.pi", "pi").replace("math.sqrt", "sqrt").replace("math.sin", "sin").replace("math.cos", "cos")
    f_expr = sp.sympify(expr_str, locals={"E": sp.E, "pi": sp.pi, "sqrt": sp.sqrt, "sin": sp.sin, "cos": sp.cos})
    x = sp.Symbol('x')
    f = sp.lambdify(x, f_expr, modules=["math"])
    b = 1
    a = -1
    n = 6
    h = (b-a)/n
    f2=sp.diff(f_expr, x, 2)
    f4=sp.diff(f_expr, x, 4)

    f2_func = sp.lambdify(x, f2, modules=["math"])
    f4_func = sp.lambdify(x, f4, modules=["math"])

    Int1 = Int_Trap(f,b,a,n)
    Int2 = Int_Simpson(f,b,a,n)
    Int3 = gauss_legendre(f,b,a,n)

    M2 = max(abs(f2_func(x)) for x in np.linspace(a, b, 200))
    M4 = max(abs(f4_func(x)) for x in np.linspace(a, b, 200))

    E_trap_teo = (b-a)/12 * h**2 * M2
    E_simp_teo = (b-a)/180 * h**4 * M4
    orden_gauss = f"≈ O(h^{2*n})"

    print(f"\n Función {z+1} Integrada por Trapecio : {Int1}, Con un Error de: ",E_trap_teo)
    print(f"\n Función {z+1} Integrada por Simpson : {Int2}, Con un Error de: ", E_simp_teo)
    print(f"\n Función {z+1} Integrada por Gauss-Legendre : {Int3}, Con un Error de: ", orden_gauss)
```
